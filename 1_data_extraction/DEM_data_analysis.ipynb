```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from LRO_data_class import LunarDataset
```

```python
ds = LunarDataset().DEMLunarData
print(ds.shape, ds.dtype)
```

```python
elevations = pd.Series(ds.ravel()[::1000])

display(elevations.describe())

print('Shape: ', ds.shape)
print('Null values')
print(np.isnan(ds).sum())
```

## Image quality

```python
sns.histplot(ds.ravel()[::1000], bins=50, kde=True)
plt.axvline(x=0, color='red', linestyle='--', label='Datum (0m)')

plt.xlabel('Elevation')
plt.ylabel('Count')
plt.title('DEM Elevation Distribution (SLDEM2015)')
plt.legend()
```
